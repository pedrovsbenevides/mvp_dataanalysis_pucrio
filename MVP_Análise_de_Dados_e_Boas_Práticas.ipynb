# 1. Problema e perguntas

## 1.1 Contexto e objetivo

A qualidade do ar é um tema relevante para a saúde pública e o meio ambiente, sendo influenciada por diferentes poluentes e condições ambientais.

Este trabalho utiliza dados de monitoramento da qualidade do ar coletados por sensores e em uma cidade italiana. O objetivo é explorar o comportamento das concentrações de poluentes, das respostas dos sensores e das condições ambientais, investigando padrões temporais e relações entre as variáveis.

A unidade de observação corresponde a um registro de medição por hora, contendo respostas dos sensores, concentrações de poluentes e variáveis ambientais.

## 1.2 Perguntas ou hipóteses

- **H1:** Os principais poluentes emitidos por veículos automotores incluem o monóxido de carbono ($CO$), dióxido de carbono ($CO_2$), óxidos de nitrogênio ($NO_x$), hidrocarbonetos ($HC$) ou compostos orgânicos voláteis ($COV$), material particulado ($MP$) e óxidos de enxofre ($SO_x$) ([CETESB](https://cetesb.sp.gov.br); [IBAMA](https://www.gov.br); [IEMA](https://energiaeambiente.org.br)).

  Dessa forma, a primeira hipótese deste trabalho é que os poluentes, emitidos por veículos, presentes no dataset ($CO$, $NO_2$ e $NO_x$), bem como as respostas de seus respectivos sensores, apresentam picos de concentração associados aos horários de pico, de tráfego mais intenso, ao longo do dia.

- **H2:** A variação da temperatura ($T$) e da umidade relativa ($RH$) interfere no comportamento e na sensibilidade dos sensores, alterando sua resposta em relação às medições de referência.

### 4.2.1 Variáveis numéricas

```python
colunas_numericas = [
    'CO(GT)', 'PT08.S1(CO)', 'NMHC(GT)', 'C6H6(GT)', 'PT08.S2(NMHC)',
    'NOx(GT)', 'PT08.S3(NOx)', 'NO2(GT)', 'PT08.S4(NO2)', 'PT08.S5(O3)',
    'T', 'RH', 'AH'
]

estatisticas = air_quality[colunas_numericas].describe().T
estatisticas['moda'] = air_quality[colunas_numericas].mode().iloc[0]

estatisticas = estatisticas[[
    'count', 'mean', 'std', 'min', '50%', 'max', 'moda'
]]

estatisticas.columns = [
    'Registros Válidos', 'Média', 'Desvio-Padrão', 'Mínimo', 'Mediana', 'Máximo', 'Moda'
]

display(estatisticas.round(2))
```

> Como pode ser observado, o valor códificado para os valores ausentes (-200) interfere significativamente nas estatísticas cálculadas e portanto precisa ser devidamente removido de modo a não interferir nos cálculos das medidas.

```python
air_quality_temp = air_quality.replace(-200, np.nan)

estatisticas = air_quality_temp[colunas_numericas].describe().T
estatisticas['moda'] = air_quality_temp[colunas_numericas].mode().iloc[0]

estatisticas = estatisticas[[
    'count', 'mean', 'std', 'min', '50%', 'max', 'moda'
]]

estatisticas.columns = [
    'Registros Válidos', 'Média', 'Desvio-Padrão', 'Mínimo', 'Mediana', 'Máximo', 'Moda'
]

display(estatisticas.round(2))
```

### 4.2.3 Análise das Estatísticas Descritivas:

**Impacto da Codificação de Ausentes:**
* A substituição do código numérico `-200` por `NaN` alterou significativamente as estatísticas descritivas em relação à tabela inicial que manteve essa codificação, revelando os valores reais das medições. Alguns impactos observados: A média de `CO(GT)` passou de `-34,21 mg/m³` na tabela inicial para `2,15 mg/m³` após a remoção do `-200`, enquanto a média de Temperatura (`T`) subiu de `9,78 °C` para `18,32 °C`.

* A análise de registros válidos, em complemento à análise de valores ausentes (seção 4.1.4), evidenciou que a variável `NMHC(GT)` possui um alto nível de incompletude com 914 registros válidos e cerca de 90,2% de dados ausentes. Será avaliado o impacto do uso dessa variável na construção das análises comparativas e se a mesma deverá ser removida.

**Poluentes de Veiculos:**
* As concentrações dos poluentes emitidos por veiculos (`CO`, `NOx` e `NO2`) e as respostas dos sensores associados exibem grande dispersão, com principal suspeita para ocorrência de picos de poluição ao longo do tempo. Para o `CO(GT)`, a mediana situa-se em `1,80 mg/m³`, mas o valor máximo atinge `11,90 mg/m³` (mais de 6 vezes a mediana). O `NOx(GT)` apresenta mediana de `180,00 ppb`, desvio-padrão elevado de `212,98 ppb` e pico máximo de `1479,00 ppb`. De forma similar, o dióxido de nitrogênio `NO2(GT)` varia de `2,00 µg/m³` a um pico de `340,00 µg/m³` (mediana de `109,00 µg/m³`). Essa alta variabilidade dos poluentes emitidos por veículos sustentam a premissa de **H1**, o que pode ser investigado visualmente em uma análise temporal por horário.

**Variação de Fatores Climáticos:**
* A Temperatura (`T`) variou de `-1,90 °C` a `44,60 °C` (média de `18,32 °C` e desvio-padrão de `8,83 °C`), enquanto a Umidade Relativa (`RH`) variou de `9,20%` a `88,70%` (média de `49,23%` e desvio-padrão de `17,32%`). Essa variabilidade climática permite testar **H2**, avaliando se as flutuações de $T$ e $RH$ alteram a resposta dos sensores em relação aos equipamentos de referência.
